# Tutorial - Units


ROSS uses [Pint](https://pint.readthedocs.io/en/stable/) for units.

If an argument is a float, ROSS assumes SI. Use `Q_(magnitude, "unit")` to attach a unit; ROSS converts it to SI internally.

Attribute values are always returned in SI. Only plot methods can change the displayed units.


```python
import numpy as np
import ross as rs

# Make sure the default renderer is set to 'notebook_connected' for inline plots in Jupyter
import plotly.io as pio

pio.renderers.default = "notebook_connected"

steel = rs.Material(name="steel", rho=7810, E=211e9, G_s=81.2e9)
rotor = rs.rotor_example()
```

# Section 1: Adding units


Import `Q_` from `ross.units` and wrap each value as `Q_(magnitude, "unit")`.

The example below creates a material with mixed Imperial and SI units.


```python
from ross.units import Q_

rho = Q_(487.56237, "lb/foot**3")  # Imperial System
E = Q_(211.0e9, "N/m**2")  # International System
G_s = Q_(81.2e9, "N/m**2")  # International System

steel4 = rs.Material(name="steel", rho=rho, E=E, G_s=G_s)
```

> **Note**: Density was given in `lb/foot**3` and is stored in SI.


```python
print(steel4)
```

The same syntax works for elements. Printed values are in SI.


## 1.1 Shaft element


```python
L = Q_(10, "in")
i_d = Q_(0.0, "meter")
o_d = Q_(0.05, "meter")

elem_pint = rs.ShaftElement(L=L, idl=i_d, odl=o_d, material=steel)
print(elem_pint)
```

## 1.2 Bearing element


```python
kxx = Q_(2.54e4, "N/in")
cxx = Q_(1e2, "N*s/m")

brg_pint = rs.BearingElement(n=0, kxx=kxx, cxx=cxx)
print(brg_pint)
```

# Section 2: Changing plot units


Plot methods take `axisname_units` arguments to convert each axis. Stored results stay in SI; only the figure changes.

Examples:

- `frequency_units`: `"rad/s"`, `"RPM"`, `"Hz"`

- `amplitude_units`: `"m"`, `"mm"`, `"in"`, `"foot"`

- `displacement_units`: `"m"`, `"mm"`, `"in"`, `"foot"`

- `rotor_length_units`: `"m"`, `"mm"`, `"in"`, `"foot"`

- `moment_units`: `"N/m"`, `"lbf/foot"`

You do not need to create elements with Pint to use these arguments. If you entered Imperial values as plain floats, ROSS still treats them as SI when converting the plot.


```python
samples = 31
speed_range = np.linspace(315, 1150, samples)

campbell = rotor.run_campbell(speed_range)
```

The default Campbell plot uses `RPM` for rotor speed (X) and natural frequency (Y).


```python
campbell.plot()
```

Pass `frequency_units` and `speed_units` to change the axes:


```python
campbell.plot(frequency_units="Hz", speed_units="RPM")
```
